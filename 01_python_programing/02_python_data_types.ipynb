## 1. Numeric classes

### 1.1. Creating numbers
Python supports three numeric types: `int` for integers, `float` for decimals, and `complex` for complex numbers. You can create each type directly using its own syntax or using a constructor.

````python
type(10)   # int
type(1,25) # float
type(1e4)  # float
type(3+4j) # complex
````

## 2. Boolean and None

### 2.1. Boolean class
Python has two Boolean values: `True` `False` (capitalized). Comparison and conditional expressions evaluate to one of these, and you combine Booleans with `and` `or`.

````python
1 == 2 # False
1 < 2 # True
'a' in 'anaconda' # True
'b' not in 'anaconda' # True
'1996'.isdigit() # True
isinstance(7., float) # True
````

#### Rules
```python
True  and True  = True
True  or  True  = True
False and False = False
False or  False = False
True  and False = False
True  or  False = True
```

````python
True and (False and True) or (False or True)
# = True and False or True
# = False or True
# = True
````

````python
sum([True, False, True, True])
````

### 3.3. Index and slicing
Python strings support slicing to create substrings. It returns a new object and the original string remains unchanged.

#### Index
Indexing rules for the string "PYTHON":

```
positive index:   0  1  2  3  4  5
string:           P  Y  T  H  O  N
negative index:  -6 -5 -4 -3 -2 -1
```

````python
my_string = 'abcdef'
my_string.index('cd')
````

#### Slicing
Use `string[start:stop:step]` for slicing, where `start:stop` defines a half-open interval $[\text{start},\text{stop})$. Omitting parameters defaults to `start=0`, `stop=len(string)`, and `step=1`.

````python
digits = '0123456789'
````

````python
digits[2:7] # 23456
digits[7:] # 789
digits[:7] # 0123456
digits[7] # 7
digits[::2] # 02468
digits[1:-1] # 12345678
digits[:7] + digits[7:] == digits # True
digits[::-1] # 9876543210
````

### 3.5. Formatting

#### Basic formatting
The `str.format()` method places arguments into `{}` placeholders:
- Empty `{}` use the next positional argument.
- `{0}`, `{1}`, etc., reference argument indices.
- `{name}` uses named parameters.

````python
'The answer is {}'.format(3+4)
````

````python
'{} and {} and {}'.format('a', 1, None)
````

````python
'{1} {0} {0} {1}'.format('three', 'seven')
````

````python
'The {cat} is {color}'.format(cat='panther', color='black')
````

You can also use `f`-strings instead of `str.format()`.

````python
cat = 'panther'
color = 'black'
f'The {cat} is {color}'
````

#### Padding and truncating

````python
# default padding uses whitespace characters and left aligns
'{:10}'.format('test')
````

````python
# right align
print('{:*>10}'.format('test'))

# left align
print('{:*<10}'.format('test'))

# center align
print('{:*^10}'.format('test'))
````

````python
# truncate to a 6-character long string
'{:.6}'.format('google chrome')
````

````python
# combine padding and truncating
'{:*>10.6}'.format('google chrome')
````

#### Number formatting

````python
# using comma (,) as thousands separator
'{x:,}'.format(x=123456789)
````

For integers, use `d`.

````python
'{: 05d}'.format(85)

# " " indicates there is a leading space " " should be used on positive number, a minus sign "-" on a negative number
# "0" indicates the number 0 is used for padding
# "5" is the fixed length
````

For floats, use `f`.

````python
# round to 2 decimal places
'{pi:.2f}'.format(pi=3.14159265359)
````

````python
# "#" option causes the output to always contain a decimal point 
'{pi:#.0f}'.format(pi=3.14159265359)
````

````python
# "#" option causes the output to always contain a decimal point 
'{pi:.2%}'.format(pi=3.14159265359)
````

### 3.6. Regular expression
A **regular expression** (regex) is a sequence of characters that defines a search pattern. For example, `^a...s$` matches any five-character string that starts with `a` and ends with `s`. Python provides the `re` module to work with regex.

#### Search patterns
Metacharacters are symbols that the regex engine interprets specially. They are: `.` `^` `$` `*` `+` `?` `[]` `()` `{}` `\` `|`.

````python
import re
````

A period `.` matches any single character except for `\\n`. The `re.findall()` function returns all substrings that match the search pattern.

````python
# return all 3-character substrings
re.findall('...', 'abcd\n1234')
````

````python
# return all 3-character substrings, including "\n"
re.findall('...', 'abcd\n1234', flags=re.DOTALL)
````

The caret `^` checks whether a string begins with specified characters. The dollar sign `$` checks whether a string ends with specified characters.

````python
# return the first 3 characters if the original string starts with "ab"
re.findall('^ab.', 'abcdefgf')
````

````python
# return the last 4 characters if the original string ends with "89"
re.findall('..89$', '0123456789')
````

Adding `flags=re.MULTILINE` instructs the regex to apply the pattern to each line instead of the entire string, capturing as much matching text as possible.

````python
string = '''
abcd
amnd
abcde
aabcdd
'''
re.findall('^a..d$', string, flags=re.MULTILINE)
````

The following metacharacters specify how many times the preceding character may repeat:
- The asterisk `*` matches **zero or more** repetitions.
- The plus sign `+` matches **one or more** repetitions.
- The question mark `?` matches **zero or one** repetition.
- The braces `{x,y}` match **at least x** and **at most y** repetitions.

````python
re.findall('ma+n', 'mn man maan maaan main')
````

````python
re.findall('ma{2,3}n', 'mn man maan maaan main')
````

The brackets `[]` define a set of characters to match. The hyphen `-` specifies a character range, and the caret `^` indicates that any character not in the set will be matched.

````python
re.findall('[Aa]', 'Anaconda')
````

````python
re.findall('[a-e]', 'jupyter notebook')
````

````python
re.findall('[^1-9]', 'abc123')
````

The backslash `\\` behaves as an escape character.
- `\d` matches any digit (equivalent to `[0-9]`). The opposite is `\D`.
- `\s` matches any whitespace character (space, newline `\n`, tab `\t`). The opposite is `\S`.
- `\w` matches any alphanumeric character (digits and letters, `[a-zA-Z0-9]`). The opposite is `\W`.

````python
re.findall('\d', 'a1b2c34')
````

````python
re.findall('\s', 'a string that spans\nmultiple lines')
````

#### Regex functions
- The `re.findall()` function returns a list of substrings that match the search pattern.
- The `re.split()` function splits the original string using the search pattern as the separators.
- The `re.sub()` function replaces all matches with the content of the `repl` parameter. The `re.subn()` function works the same but returns both the replaced string and the number of substitutions made.
- The `re.search()` and `re.match()` functions both return a `re.match` object.

````python
import re
````

````python
re.findall('\d+', 'date:08month:12year:2019')
````

````python
re.split(':\d+', 'date:08month:12year:2019', maxsplit=2)
````

````python
re.sub(
    pattern=r'\s+',
    repl='_',
    string='tiger  lion cheetah \n leopard \t jaguar'
)
````

````python
re.subn(
    pattern=r'\s+',
    repl='_',
    string='tiger  lion cheetah \n leopard \t jaguar'
)
````

````python
if re.match('...', 'abcdef'):
    print(True)
else:
    print(False)
````

````python
# return 2 subgroups
match = re.search(r'(\d{3}) (\d{2})', '39801 356, 2102 1111')
print(match.group())
print(match.group(1))
print(match.groups())
````

````python
# return the indices of the first and the last character of the match
match = re.search(r'\d+', 'abc12345de')
print(match.start())
print(match.end())
print(match.span())
````

### 4.2. Timedelta
The difference between two `dt.datetime` objects yields a `dt.timedelta` object. However, since this class doesn’t support adding intervals longer than one day, it’s better to use the [`dateutil.relativedelta`] module instead.

[`dateutil.relativedelta`]: https://dateutil.readthedocs.io/en/stable/relativedelta.html

````python
import datetime as dt
import dateutil.relativedelta as du
````

````python
date1 = dt.date(1945, 9, 2)
date2 = dt.date.today()
date2 - date1
````

````python
du.relativedelta(date2, date1)
````

````python
# calculating with various frequencies
date1 + du.relativedelta(years=5, months=-3, weeks=1, days=-10)
````

````python
# substitue with a specific day
date1 + du.relativedelta(day=31)
````

````python
# next 2 Mondays
date1 + du.relativedelta(weekday=du.MO(2))
````

````python
# date1 is already a Sunday
date1 + du.relativedelta(weekday=du.SU, days=1)
````

:::{note}
- Substituting day 31 explicitly always returns the last day of that month.  
- Because 1945-09-02 is already a Sunday, `dateutil` treats “next Sunday” as the same day. To advance to the following Sunday, add one day.  
- When using `dateutil`, avoid mixing too many date/time frequency types.
:::

### 4.3. Datetime formatting
- `strftime()` **formats** `datetime` objects into strings.
- `strptime()` **parses** strings to construct `datetime` objects.

#### Format codes
Code                                  |Meaning                         |Example                             |
:-------------------------------------|:-------------------------------|:-----------------------------------|
`%y`|Year, 2 last digits             |00, 01, 02,..., 99                  |
`%Y`|Year, full                      |0001, 0002,..., 2019, 2020,..., 9999|
`%b`|Month name, abbreviated         |Jan, Feb,..., Dec                   |
`%B`|Month name, full                |January, February,..., December     |
`%m`|Month of the year, zero-padded  |01, 02, 03,..., 12                  |
`%W`|Week of the year, zero-padded   |00, 01, 02,..., 53                  |
`%d`|Day of the month, zero-padded   |01, 02, 03,..., 31                  |
`%a`|Weekday name, abbreviated       |Mon, Tue,..., Sun                   |
`%A`|Weekday name, full              |Monday, Tuesday,..., Sunday         |
`%H`|Hour, 24-hour clock, zero-padded|00, 01, 02,..., 23                  |
`%I`|Hour, 12-hour clock, zero-padded|01, 02, 03,..., 12                  |
`%p`|AM or PM                        |AM, PM                              |
`%M`|Minute, zero-padded             |00, 01, 02,..., 59                  |
`%S`|Second, zero-padded             |00, 01, 02,..., 59                  |

````python
import datetime as dt
````

````python
# strptime() converts a string to a datetime object
string_1 = '11/8/2018 12:18:00 PM'
datetime_1 = dt.datetime.strptime(string_1, '%m/%d/%Y %I:%M:%S %p')
datetime_1
````

````python
# strftime() converts a datetime object to a string
datetime_2 = dt.datetime(2018, 11, 8, 12, 18)
string_2 = datetime_2.strftime('%m/%d/%Y %I:%M:%S %p')
string_2
````

````python
# get date, day, month of the current day, print as "DD-MM-YYYY"
date = dt.date.today()
date.strftime('%d-%m-%Y')
````
