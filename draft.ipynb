```python
from mlkit import np, plt

f = lambda x: -0.4 * x**2 + 4
f_prime = lambda x: -0.8 * x

a, x1, x2 = -2.0, -0.5, 1.0
xr = np.linspace(-4, 5, 400)

fig, ax = plt.subplots(figsize=(6,4))
ax.plot(xr, f(xr), color='gray', lw=3, zorder=1)

lines = [('royalblue', a, True), ('indianred', x1, False), ('darkorange', x2, False)]

for col, px, is_tangent in lines:
    m = f_prime(a) if is_tangent else (f(px) - f(a)) / (px - a)
    ax.plot(xr, m * (xr - a) + f(a), color=col, lw=2, alpha=0.9)
    
    py = f(px)
    ax.plot([px, px], [0, py], 'gray', ls=':', lw=2)
    ax.scatter(px, py, color=col, s=130, zorder=5, edgecolors='white')
    ax.scatter(px, 0, color='dimgray', s=60, zorder=5)
    ax.text(px, -0.4, '$a$' if is_tangent else '$x$', fontsize=14, ha='center', weight='bold')

ax.axhline(0, color='black', lw=1.5)
ax.axvline(0.2, color='black', lw=1.5)
ax.set(xlim=(-4, 4), ylim=(-1, 6))
ax.axis('off')
plt.show()
```
