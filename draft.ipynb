```python
import sys; sys.path.append('..')
from mlkit import plt, np

fig, ax = plt.subplots(figsize=(10, 7))

b_pts = np.array([[1, 5], [2, 6], [3, 6], [1.7, 4.3], [1.5, 3.5], [2.4, 2.6]])
r_pts = np.array([[5, 3], [6, 2], [4.5, 1.5], [5.5, 0.8], [4.3, 3.7]])
sv_b = b_pts[[3, 5]]
sv_r = r_pts[[4]]

ax.scatter(b_pts[:, 0], b_pts[:, 1], c='#3b5998', s=300, zorder=3)
ax.scatter(r_pts[:, 0], r_pts[:, 1], c='#b22222', s=300, zorder=3)
ax.scatter(sv_b[:, 0], sv_b[:, 1], s=700, facecolors='none', edgecolors='#333', lw=2, zorder=4)
ax.scatter(sv_r[:, 0], sv_r[:, 1], s=700, facecolors='none', edgecolors='#333', lw=2, zorder=4)

x = np.linspace(0, 8, 100)
ax.plot(x, x - 0.5, 'k-', lw=2, zorder=2)
ax.plot(x, x + 1.2, 'k:', lw=3, zorder=2)
ax.plot(x, x - 2.2, 'k:', lw=3, zorder=2)

ax.annotate('', xy=(1.5, 2.7), xytext=(2.2, 2.0), arrowprops=dict(arrowstyle='<->', lw=1.5))
ax.annotate('', xy=(0.8, 2.0), xytext=(1.5, 1.3), arrowprops=dict(arrowstyle='<->', lw=1.5))
ax.text(1.8, 2.3, 'margin', rotation=315, ha='center')
ax.text(1.1, 1.6, 'margin', rotation=315, ha='center')

ax.text(3.5, 5.0, '$w_0 + w_1x_1 + w_2x_2 = 1$', rotation=45)
ax.text(5.0, 4.8, '$w_0 + w_1x_1 + w_2x_2 = 0$', rotation=45)
ax.text(6.5, 4.6, '$w_0 + w_1x_1 + w_2x_2 = -1$', rotation=45)

ax.text(7.5, 4.0, 'support\nvectors', fontsize=12, fontweight='bold')
for p in [sv_b[0], sv_b[1], sv_r[0]]:
    ax.annotate('', xy=p, xytext=(7.4, 4.2), arrowprops=dict(arrowstyle='->', ls=':', lw=0.5))

ax.set_xlim(0, 9); ax.set_ylim(0, 8)
ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')
for s in ['top', 'right']: ax.spines[s].set_visible(False)
ax.plot(1, 0, ">k", transform=ax.get_yaxis_transform(), clip_on=False)
ax.plot(0, 1, "^k", transform=ax.get_xaxis_transform(), clip_on=False)
ax.set_xticks([]); ax.set_yticks([])
ax.set_xlabel('$x_1$', loc='right', fontsize=14); ax.set_ylabel('$x_2$', loc='top', rotation=0, fontsize=14)

plt.tight_layout()
plt.show()
```

```python
import sys; sys.path.append('..')
from mlkit import plt, np

fig, ax = plt.subplots(figsize=(8,5))

# 1. Define Points with wider separation
# Support Vectors (adjusted to widen the gap)
sv_b = np.array([[2.2, 3.2]])   # Blue SV (Intercept b ~ 1.0)
sv_r = np.array([[4.8, 2.8]])   # Red SV (Intercept b ~ -2.0)

# Other cluster points (pushed away from margin)
b_pts = np.array([[1.0, 5.0], [2.0, 5.5], [0.8, 4.2], [1.5, 4.0], [2.2, 3.2]])
r_pts = np.array([[6.0, 1.5], [5.5, 0.5], [6.5, 2.0], [5.2, 1.8], [4.8, 2.8]])

# 2. Plot Scatters
ax.scatter(b_pts[:-1, 0], b_pts[:-1, 1], c='#3b5998', s=300, zorder=3)
ax.scatter(r_pts[1:, 0], r_pts[1:, 1], c='#b22222', s=300, zorder=3)

# Highlight Support Vectors
ax.scatter(sv_b[:, 0], sv_b[:, 1], s=300, c='#3b5998', zorder=3)
ax.scatter(sv_r[:, 0], sv_r[:, 1], s=300, c='#b22222', zorder=3)
ax.scatter(sv_b[:, 0], sv_b[:, 1], s=700, facecolors='none', edgecolors='#333', lw=2, zorder=4)
ax.scatter(sv_r[:, 0], sv_r[:, 1], s=700, facecolors='none', edgecolors='#333', lw=2, zorder=4)

# 3. Plot Lines (Slope m=1)
# Top: hits (2.2, 3.2) -> 3.2 = 2.2 + b -> b = 1.0
# Bot: hits (4.8, 2.8) -> 2.8 = 4.8 + b -> b = -2.0
# Mid: b = -0.5
x = np.linspace(0, 9, 100)
ax.plot(x, x - 0.5, 'k-', lw=2, zorder=2)       # Midline
ax.plot(x, x + 1.0, 'k:', lw=2, zorder=2)       # Upper Margin
ax.plot(x, x - 2.0, 'k:', lw=2, zorder=2)       # Lower Margin

# 4. Margin Arrows (Wider now)
# Perpendicular slope = -1. Start at mid (x=1.8, y=1.3) -> Up/Down
ax.annotate('', xy=(1.3, 2.3), xytext=(1.8, 1.8), arrowprops=dict(arrowstyle='<->', lw=1.5))
ax.annotate('', xy=(2.3, 1.3), xytext=(1.8, 1.8), arrowprops=dict(arrowstyle='<->', lw=1.5))
ax.text(1.35, 2.2, 'margin', rotation=315, ha='right', va='bottom', fontsize=11)
ax.text(2.25, 1.4, 'margin', rotation=315, ha='left', va='top', fontsize=11)

# 5. Equation Text (Re-aligned to new lines)
rot = 45
ax.text(3.1, 4.3, '$w_0 + w_1x_1 + w_2x_2 = 1$', rotation=rot, fontsize=12)
ax.text(4.3, 4.1, '$w_0 + w_1x_1 + w_2x_2 = 0$', rotation=rot, fontsize=12)
ax.text(5.5, 3.7, '$w_0 + w_1x_1 + w_2x_2 = -1$', rotation=rot, fontsize=12)

# 6. Support Vector Labels
ax.text(7.5, 4.2, 'support\nvectors', fontsize=12, fontweight='bold')
ax.annotate('', xy=sv_b[0], xytext=(7.4, 4.4), arrowprops=dict(arrowstyle='->', ls=':', lw=0.8))
ax.annotate('', xy=sv_r[0], xytext=(7.4, 4.4), arrowprops=dict(arrowstyle='->', ls=':', lw=0.8))

# 7. Final Formatting
ax.set(xlim=(0, 9), ylim=(0, 7), aspect='equal')
ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')
for s in ['top', 'right']: ax.spines[s].set_visible(False)
ax.plot(1, 0, ">k", transform=ax.get_yaxis_transform(), clip_on=False)
ax.plot(0, 1, "^k", transform=ax.get_xaxis_transform(), clip_on=False)
ax.set_xticks([]); ax.set_yticks([])
ax.set_xlabel('$x_1$', loc='right', fontsize=14)
ax.set_ylabel('$x_2$', loc='top', rotation=0, fontsize=14)

plt.tight_layout()
plt.show()
```
