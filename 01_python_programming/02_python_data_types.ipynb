# Python: Data Types

## 1. Numeric types

### 1.1. Creating numbers
Python has three built-in numeric types: `int` for integers, `float` for floating-point numbers, and `complex` for complex numbers. You can create numbers using numeric literals or the corresponding constructors: `int()`, `float()`, and `complex()`.

````python
type(10)   # int
type(1.25) # float
type(1e4)  # float
type(3+4j) # complex
````

````python
int(3.14)          # 3
int('3f', base=16) # 63

float(5)     # 5.0
float('inf') # inf
float('nan') # nan

complex(7)    # 7+0j
complex(3, 4) # 3+4j
````

### 1.2. Arithmetic operators

````python
5 + 3.14 # sum
6 - 9 # difference
3 * (5+2.5j) # product
40 / 8 # quotient
41 // 5 # floor quotient
41 % 5 # remainder
2 ** 5 # exponentiation
````

````python
dividend = 41
divisor = 5
quotient, remainder = divmod(dividend, divisor)
print(f'{dividend} = {divisor} * {quotient} + {remainder}')
````

````python
from math import sqrt
sqrt(25)
````

````python
abs(5 + 12j)       # 13.0
round(133.45, -1)  # 130.0
pow(5, 2)          # 25
````

## 2. Boolean and None

### 2.1. The Boolean type
Python has two Boolean values: `True` and `False`. Comparisons and membership tests commonly produce Boolean values, which can be combined using the `and`, `or`, and `not` operators and used in conditional statements.

````python
bool(0)        # False
bool('')       # False
bool(None)     # False
bool(1)        # True
bool('False')  # True
````

````python
1 == 2                   # False: equality comparison
1 < 2                    # True: ordering comparison
'a' in 'anaconda'        # True: membership test
'b' not in 'anaconda'    # True: negative membership test
'1996'.isdigit()         # True: string predicate
isinstance(7.0, float)   # True: type check
````

#### Boolean operators
- `and` returns `True` only when both operands are true.  
- `or` returns `True` when at least one operand is true.  
- `not` reverses the truth value of its operand.
- Because `bool` is a subtype of `int`, `True` behaves like `1` and `False` behaves like `0` in numeric operations. Therefore, summing Boolean values counts the number of `True` values.

````python
has_account = True
has_password = False
uses_single_sign_on = True

can_log_in = has_account and (has_password or uses_single_sign_on)
can_log_in  # True
````

````python
sum([True, False, True, True])
````

### 2.2. The `None` value
`None` is a special value that represents the absence of a value. It is commonly used as a default value or as the return value of functions that do not explicitly return another object. `None` is distinct from `0`, `False`, and an empty string. Use the identity operators `is` and `is not` when checking whether a value is `None`.

````python
result = None
type(result)
````

````python
None == 0      # False
None is None   # True
````

## 3. Strings
A Python string is an immutable sequence of Unicode characters used to represent text.

### 3.1. String manipulation

#### Creating strings
You can create string literals using single quotes `''`, double quotes `""`, or triple quotes. Triple-quoted strings can span multiple lines. You can also use the `str()` constructor to obtain the string representation of another object.

````python
'Data Science'
````

````python
print('''This string
spans multiple lines''')
````

````python
str(245)
````

#### Concatenation and repetition

````python
label = 'Python'
message = label + ' is useful.'   # concatenation

separator = '-' * 20   # repetition
````

#### Functions work with strings
Because strings are sequences, functions such as `len()`, `min()`, and `max()` can operate on them. The `min()` and `max()` functions compare individual characters according to their Unicode ordering.

````python
text = 'Python'

len(text)  # 6
min(text)  # 'P'
max(text)  # 'y'
````

#### String methods

````python
'jack.o.lantern'.split( sep='.')            # ['jack', 'o', 'lantern']
'-'.join(['jack', 'o', 'lantern'])          # 'jack-o-lantern'
'old string old'.replace('old', 'new', 1)   # 'new string old'
````

````python
text = 'jupyter noteBOOK'

text.title()  # 'Jupyter Notebook'
text.upper()  # 'JUPYTER NOTEBOOK'
text.lower()  # 'jupyter notebook'
````

````python
'notebook'.startswith('no')  # True
'1234'.isdecimal()           # True
'abc'.isalpha()              # True
'ab12'.isalnum()             # True
'Python Basics'.istitle()    # True
````

### 3.2. Indexing and slicing
You can access individual characters by index and create substrings by slicing. Because strings are immutable, slicing returns a new string and leaves the original string unchanged.

#### Indexing
Indexing rules for the string "PYTHON":

```
positive index:   0  1  2  3  4  5
string:           P  Y  T  H  O  N
negative index:  -6 -5 -4 -3 -2 -1
```

````python
text = 'PYTHON'

text[3]            # H
text[-1]           # N
text.index('TH')   # 2
````

#### Slicing
Use `string[start:stop:step]` to extract a slice. Here are the key behaviours:
- The character at `start` is included, while the character at `stop` is excluded. The default step is `1`.
- When the step is positive, omitted bounds normally select from the beginning or through the end of the string.
- A negative step traverses the string from right to left. For example, `string[::-1]` reverses a string.

````python
digits = '0123456789'

digits[2:7]                # '23456'
digits[7:]                 # '789'
digits[:7]                 # '0123456'
digits[7]                  # '7'
digits[::2]                # '02468'
digits[1:-1]               # '12345678'
digits[:7] + digits[7:] == digits  # True
digits[::-1]               # '9876543210'
````

### 3.3. String prefixes and escape sequences

#### Escape sequences
A backslash begins an escape sequence inside a regular string literal. Common examples include `\n` for a newline, `\t` for a tab, `\\` for a literal backslash, and `\'` or `\"` for quote characters. Unicode characters can be written using `\u` followed by four hexadecimal digits or `\U` followed by eight hexadecimal digits.

````python
# new line
print('aaa\nbbb')
````

````python
print('It\'s raining')
print("It's raining") # often clearer
````

````python
# Greek letters, using Unicode
print('''alpha: \u03b1 \t beta: \u03b2 \t gamma: \u03b3
delta: \u03b4 \t epsilon: \u03b5 \t theta: \u03b8
lambda: \u03bb \t mu: \u03bc \t\t pi: \u03c0
sigma: \u03c3 \t phi: \u03c6 \t omega: \u03c9''')
````

#### Bytes
A `bytes` object represents binary data as an immutable sequence of byte values from 0 to 255. An ASCII-only bytes literal can be written with the `b` prefix. To convert Unicode text to bytes, encode it using a character encoding such as UTF-8.

````python
text = 'Tiếng Việt'
encoded = text.encode('utf-8')
decoded = encoded.decode('utf-8')

encoded
# b'Ti\xe1\xba\xbfng Vi\xe1\xbb\x87t'

decoded
# 'Tiếng Việt'
````

#### Raw string
A raw string literal uses the `r` prefix and leaves backslashes mostly unprocessed. Raw strings are useful for regular-expression patterns and Windows-style paths.

````python
print(r'a\nb')
````

````python
r'C:\Users'
````

#### Formatted string
Prefix a string literal with `f` to create an f-string. Expressions inside braces are evaluated and inserted into the resulting string.

````python
f'1 + 2 = {1+2}'
````

````python
name = 'Ben'
f'My name is {name}'
````

### 3.4. Formatting

#### Basic formatting
The `str.format()` method inserts values into replacement fields marked by braces:
- `{}` uses the next positional argument.
- `{0}`, `{1}`, and similar fields refer to positional argument indices.
- `{name}` refers to a named argument.

````python
# next arguments
template = '{} scored {} points. Congratulations, {}!'
template.format('Ben', 95, 'Ben')

# positional arguments
template = '{0} scored {1} points. Congratulations, {0}!'
template.format('Ben', 95)

# named argument
template = '{name} scored {score} points. Congratulations, {name}!'
template.format(name='Ben', score=95)

# f-string, more readable
name = 'Ben'
score = 95
f'{name} scored {score} points. Congratulations, {name}!'
````

#### Padding and truncating

````python
# Strings are left-aligned by default and padded with spaces when a field width is specified
f'{42:10}'     # '        42'
f'{42:*>10}'   # '********42'
f'{42:*<10}'   # '42********'
f'{42:*^10}'   # '****42****'
````

````python
# limit the formatted value to at most six characters
'{:.6}'.format('google chrome')
````

````python
# combine padding and truncating
'{:*>10.6}'.format('google chrome')
````

#### Number formatting
Use `d`, `f`, and `%` in number format specifiers to display integers, fixed-point decimals, and percentages.

````python
# thousands separators
'{x:,}'.format(x=123456789)   # 123,456,789
````

````python
# The `0` requests zero padding, and `5` specifies a total field width of five characters
f'{85:05d}'   # '00085'
f'{-85:05d}'  # '-0085'
````

````python
# `.2` round to 2 decimal places
# `#` option causes the output to always contain a decimal point

pi = 3.14159265359

f'{pi:.2f}'   # '3.14'
f'{pi:#.0f}'  # '3.'
````

````python
rate = 0.314159
f'{rate:.2%}'  # '31.42%'
````

## 4. Regular expression
A regular expression (regex) is a sequence of characters that defines a search pattern. For example, `^a...s$` matches any five-character string that starts with `a` and ends with `s`. Python provides the `re` module to work with regex.

#### Search patterns
Metacharacters are symbols that the regex engine interprets specially. They are: `.` `^` `$` `*` `+` `?` `[]` `()` `{}` `\` `|`.

````python
import re
````

A period `.` matches any single character except for `\\n`. The `re.findall()` function returns all substrings that match the search pattern.

````python
# return all 3-character substrings
re.findall('...', 'abcd\n1234')
````

````python
# return all 3-character substrings, including "\n"
re.findall('...', 'abcd\n1234', flags=re.DOTALL)
````

The caret `^` checks whether a string begins with specified characters. The dollar sign `$` checks whether a string ends with specified characters.

````python
# return the first 3 characters if the original string starts with "ab"
re.findall('^ab.', 'abcdefgf')
````

````python
# return the last 4 characters if the original string ends with "89"
re.findall('..89$', '0123456789')
````

Adding `flags=re.MULTILINE` instructs the regex to apply `^` and `$` patterns to each line instead of the entire string, capturing as much matching text as possible.

````python
string = '''
abcd
amnd
abcde
aabcdd
'''
re.findall('^a..d$', string, flags=re.MULTILINE)
````

The following metacharacters specify how many times the preceding character may repeat:
- The asterisk `*` matches zero or more repetitions.
- The plus sign `+` matches one or more repetitions.
- The question mark `?` matches zero or one repetition.
- The braces `{x,y}` match at least x and at most y repetitions.

````python
re.findall('ma+n', 'mn man maan maaan main')
````

````python
re.findall('ma{2,3}n', 'mn man maan maaan main')
````

The brackets `[]` define a set of characters to match. The hyphen `-` specifies a character range, and the caret `^` indicates that any character not in the set will be matched.

````python
re.findall('[Aa]', 'Anaconda')
````

````python
re.findall('[a-e]', 'jupyter notebook')
````

````python
re.findall('[^1-9]', 'abc123')
````

The backslash `\\` behaves as an escape character.
- `\d` matches any digit (equivalent to `[0-9]`). The opposite is `\D`.
- `\s` matches any whitespace character (space, newline `\n`, tab `\t`). The opposite is `\S`.
- `\w` matches any alphanumeric character (digits and letters, `[a-zA-Z0-9]`). The opposite is `\W`.

````python
re.findall('\d', 'a1b2c34')
````

````python
re.findall('\s', 'a string that spans\nmultiple lines')
````

#### Regex functions
- The `re.findall()` function returns a list of substrings that match the search pattern.
- The `re.split()` function splits the original string using the search pattern as the separators.
- The `re.sub()` function replaces all matches with the content of the `repl` parameter. The `re.subn()` function works the same but returns both the replaced string and the number of substitutions made.
- The `re.search()` and `re.match()` functions both return a `re.match` object.

````python
import re
````

````python
re.findall('\d+', 'date:08month:12year:2019')
````

````python
re.split(':\d+', 'date:08month:12year:2019', maxsplit=2)
````

````python
re.sub(
    pattern=r'\s+',
    repl='_',
    string='tiger  lion cheetah \n leopard \t jaguar'
)
````

````python
re.subn(
    pattern=r'\s+',
    repl='_',
    string='tiger  lion cheetah \n leopard \t jaguar'
)
````

````python
if re.match('...', 'abcdef'):
    print(True)
else:
    print(False)
````

````python
# return 2 subgroups
match = re.search(r'(\d{3}) (\d{2})', '39801 356, 2102 1111')
print(match.group())
print(match.group(1))
print(match.groups())
````

````python
# return the indices of the first and the last character of the match
match = re.search(r'\d+', 'abc12345de')
print(match.start())
print(match.end())
print(match.span())
````

### 5.2. Timedelta
The difference between two `dt.datetime` objects yields a `dt.timedelta` object. However, since this class doesn’t support adding intervals with variable duration (like months and years), it’s better to use the [`dateutil.relativedelta`] module instead.

[`dateutil.relativedelta`]: https://dateutil.readthedocs.io/en/stable/relativedelta.html

````python
import datetime as dt
import dateutil.relativedelta as du
````

````python
date1 = dt.date(1945, 9, 2)
date2 = dt.date.today()
date2 - date1
````

````python
du.relativedelta(date2, date1)
````

````python
# calculating with various frequencies
date1 + du.relativedelta(years=5, months=-3, weeks=1, days=-10)
````

````python
# substitue with a specific day
date1 + du.relativedelta(day=31)
````

````python
# next 2 Mondays
date1 + du.relativedelta(weekday=du.MO(2))
````

````python
# date1 is already a Sunday
date1 + du.relativedelta(weekday=du.SU, days=1)
````

:::{note}
- Substituting day 31 explicitly always returns the last day of that month.
- Because 1945-09-02 is already a Sunday, `dateutil` treats “next Sunday” as the same day. To advance to the following Sunday, add one day.
- When using `dateutil`, avoid mixing too many date/time frequency types.
:::

### 5.3. Datetime formatting
- `strftime()` formats `datetime` objects into strings.
- `strptime()` parses strings to construct `datetime` objects.

#### Format codes
Code                                  |Meaning                         |Example                             |
:-------------------------------------|:-------------------------------|:-----------------------------------|
`%y`|Year, 2 last digits             |00, 01, 02,..., 99                  |
`%Y`|Year, full                      |0001, 0002,..., 2019, 2020,..., 9999|
`%b`|Month name, abbreviated         |Jan, Feb,..., Dec                   |
`%B`|Month name, full                |January, February,..., December     |
`%m`|Month of the year, zero-padded  |01, 02, 03,..., 12                  |
`%W`|Week of the year, zero-padded   |00, 01, 02,..., 53                  |
`%d`|Day of the month, zero-padded   |01, 02, 03,..., 31                  |
`%a`|Weekday name, abbreviated       |Mon, Tue,..., Sun                   |
`%A`|Weekday name, full              |Monday, Tuesday,..., Sunday         |
`%H`|Hour, 24-hour clock, zero-padded|00, 01, 02,..., 23                  |
`%I`|Hour, 12-hour clock, zero-padded|01, 02, 03,..., 12                  |
`%p`|AM or PM                        |AM, PM                              |
`%M`|Minute, zero-padded             |00, 01, 02,..., 59                  |
`%S`|Second, zero-padded             |00, 01, 02,..., 59                  |

````python
import datetime as dt
````

````python
# strptime() converts a string to a datetime object
string_1 = '11/8/2018 12:18:00 PM'
datetime_1 = dt.datetime.strptime(string_1, '%m/%d/%Y %I:%M:%S %p')
datetime_1
````

````python
# strftime() converts a datetime object to a string
datetime_2 = dt.datetime(2018, 11, 8, 12, 18)
string_2 = datetime_2.strftime('%m/%d/%Y %I:%M:%S %p')
string_2
````

````python
# get date, day, month of the current day, print as "DD-MM-YYYY"
date = dt.date.today()
date.strftime('%d-%m-%Y')
````
