# Python: Strings

## 1. String foundations
A Python string is an immutable sequence of Unicode characters used to represent text.

### 1.1. String manipulation

#### Creating strings
You can create string literals using single quotes `''`, double quotes `""`, or triple quotes. Triple-quoted strings can span multiple lines. You can also use the `str()` constructor to obtain the string representation of another object.

````python
'Data Science'
````

````python
print('''This string
spans multiple lines''')
````

````python
str(245)
````

#### Concatenation and repetition

````python
label = 'Python'
message = label + ' is useful.'   # concatenation

separator = '-' * 20   # repetition
````

#### Functions work with strings
Because strings are sequences, functions such as `len()`, `min()`, and `max()` can operate on them. The `min()` and `max()` functions compare individual characters according to their Unicode ordering.

````python
text = 'Python'

len(text)  # 6
min(text)  # 'P'
max(text)  # 'y'
````

#### String methods

````python
'jack.o.lantern'.split( sep='.')            # ['jack', 'o', 'lantern']
'-'.join(['jack', 'o', 'lantern'])          # 'jack-o-lantern'
'old string old'.replace('old', 'new', 1)   # 'new string old'
````

````python
text = 'jupyter noteBOOK'

text.title()  # 'Jupyter Notebook'
text.upper()  # 'JUPYTER NOTEBOOK'
text.lower()  # 'jupyter notebook'
````

````python
'notebook'.startswith('no')  # True
'1234'.isdecimal()           # True
'abc'.isalpha()              # True
'ab12'.isalnum()             # True
'Python Basics'.istitle()    # True
````

### 1.2. Indexing and slicing
You can access individual characters by index and create substrings by slicing. Because strings are immutable, slicing returns a new string and leaves the original string unchanged.

#### Indexing
Indexing rules for the string "PYTHON":

```
positive index:   0  1  2  3  4  5
string:           P  Y  T  H  O  N
negative index:  -6 -5 -4 -3 -2 -1
```

````python
text = 'PYTHON'

text[3]            # H
text[-1]           # N
text.index('TH')   # 2
````

#### Slicing
Use `string[start:stop:step]` to extract a slice. Here are the key behaviours:
- The character at `start` is included, while the character at `stop` is excluded. The default step is `1`.
- When the step is positive, omitted bounds normally select from the beginning or through the end of the string.
- A negative step traverses the string from right to left. For example, `string[::-1]` reverses a string.

````python
digits = '0123456789'

digits[2:7]                # '23456'
digits[7:]                 # '789'
digits[:7]                 # '0123456'
digits[7]                  # '7'
digits[::2]                # '02468'
digits[1:-1]               # '12345678'
digits[:7] + digits[7:] == digits  # True
digits[::-1]               # '9876543210'
````

### 1.3. String prefixes and escape sequences

#### Escape sequences
A backslash begins an escape sequence inside a regular string literal. Common examples include `\n` for a newline, `\t` for a tab, `\\` for a literal backslash, and `\'` or `\"` for quote characters. Unicode characters can be written using `\u` followed by four hexadecimal digits or `\U` followed by eight hexadecimal digits.

````python
# new line
print('aaa\nbbb')
````

````python
print('It\'s raining')
print("It's raining") # often clearer
````

````python
# Greek letters, using Unicode
print('''alpha: \u03b1 \t beta: \u03b2 \t gamma: \u03b3
delta: \u03b4 \t epsilon: \u03b5 \t theta: \u03b8
lambda: \u03bb \t mu: \u03bc \t\t pi: \u03c0
sigma: \u03c3 \t phi: \u03c6 \t omega: \u03c9''')
````

#### Bytes
A `bytes` object represents binary data as an immutable sequence of byte values from 0 to 255. An ASCII-only bytes literal can be written with the `b` prefix. To convert Unicode text to bytes, encode it using a character encoding such as UTF-8.

````python
text = 'Tiếng Việt'
encoded = text.encode('utf-8')
decoded = encoded.decode('utf-8')

encoded
# b'Ti\xe1\xba\xbfng Vi\xe1\xbb\x87t'

decoded
# 'Tiếng Việt'
````

#### Raw string
A raw string literal uses the `r` prefix and leaves backslashes mostly unprocessed. Raw strings are useful for regular-expression patterns and Windows-style paths.

````python
print(r'a\nb')
````

````python
r'C:\Users'
````

#### Formatted string
Prefix a string literal with `f` to create an f-string. Expressions inside braces are evaluated and inserted into the resulting string.

````python
f'1 + 2 = {1+2}'
````

````python
name = 'Ben'
f'My name is {name}'
````

### 1.4. Formatting

#### Basic formatting
The `str.format()` method inserts values into replacement fields marked by braces:
- `{}` uses the next positional argument.
- `{0}`, `{1}`, and similar fields refer to positional argument indices.
- `{name}` refers to a named argument.

````python
# next arguments
template = '{} scored {} points. Congratulations, {}!'
template.format('Ben', 95, 'Ben')

# positional arguments
template = '{0} scored {1} points. Congratulations, {0}!'
template.format('Ben', 95)

# named argument
template = '{name} scored {score} points. Congratulations, {name}!'
template.format(name='Ben', score=95)

# f-string, more readable
name = 'Ben'
score = 95
f'{name} scored {score} points. Congratulations, {name}!'
````

#### Padding and truncating
Strings are left-aligned by default and padded with spaces when a field width is specified.

````python

f'{42:10}'     # '        42'
f'{42:*>10}'   # '********42'
f'{42:*<10}'   # '42********'
f'{42:*^10}'   # '****42****'
````

````python
# limit the formatted value to at most six characters
'{:.6}'.format('google chrome')
````

````python
# combine padding and truncating
'{:*>10.6}'.format('google chrome')
````

#### Number formatting
Use `d`, `f`, and `%` in number format specifiers to display integers, fixed-point decimals, and percentages.

````python
# thousands separators
'{x:,}'.format(x=123456789)   # 123,456,789
````

````python
# `0` requests zero padding
# `5` specifies a total field width of five characters
f'{85:05d}'   # '00085'
f'{-85:05d}'  # '-0085'
````

````python
# `.2` round to 2 decimal places
# `#` option causes the output to always contain a decimal point

pi = 3.14159265359

f'{pi:.2f}'   # '3.14'
f'{pi:#.0f}'  # '3.'
````

````python
rate = 0.314159
f'{rate:.2%}'  # '31.42%'
````

## 2. Regular expressions
A regular expression, or regex, is a pattern used to search, extract, split, or replace text. Python provides regular-expression support through the built-in `re` module. For example, the pattern `^a...s$` matches a five-character string that starts with `a` and ends with `s`.

:::{note}
- Regex patterns often contain backslashes `\`. Write them as raw strings, with an `r` prefix, so that Python treats the backslashes literally, for example: `r'^a...s$'`.
- Regex is not a universal language; Python's regex syntax and behavior can differ from those of other programming languages.
:::

### 2.1. Basic patterns

#### Metacharacters

| Regex syntax | Matches                                          | Negated form |
| :----------- | :----------------------------------------------- | :----------- |
| `.`          | Any single character except a newline            |              |
| `\.`         | A literal dot `.`                                |              |
| `a\|b`       | Either a or b                                    |              |
| `\d`         | Any digit                                        | `\D`         |
| `\w`         | Any word character: letter, digit, or underscore | `\W`         |
| `\s`         | Any whitespace character                         | `\S`         |

#### Character classes
Square brackets `[]` define a character class, which matches any one character listed inside the brackets or included in a specified range. Placing a caret `^` immediately after the opening bracket negates the class, causing it to match any character not listed.

| Regex syntax  | Matches                                    | Negated form   |
| :------------ | :----------------------------------------- | :------------- |
| `[abc]`       | Any of the characters a, b, or c           | `[^abc]`       |
| `[a-z]`       | Any lowercase letter from a to z           | `[^a-z]`       |
| `[0-9]`       | Any digit from 0 to 9                      | `[^0-9]`       |
| `[A-Cx-z3-5]` | Any character of A, B, C, x, y, z, 3, 4, 5 | `[^A-Cx-z3-5]` |

#### Quantifiers
Quantifiers are placed immediately after a character, character class, or group. They specify how many times the preceding element may occur.

| Regex syntax | Matches                         |
| :----------- | :------------------------------ |
| `*`          | Zero or more occurrences        |
| `+`          | One or more occurrences         |
| `?`          | Zero or one occurrence          |
| `{n}`        | Exactly `n` occurrences         |
| `{n,}`       | At least `n` occurrences        |
| `{n,m}`      | Between `n` and `m` occurrences |

For example:
- `a+` matches `a`, `aa`, `aaa`, etc.
- `a{3,4}` matches `aaa` and `aaaa`

:::{note}
By default, quantifiers are *greedy*, so they match as much text as possible. For example, `<.*>` applied to `<p>Hello</p><p>World</p>` matches the entire string because `.*` expands to the last `>`.

Adding `?` makes a quantifier *lazy*: `*?`, `+?`, `??`, or `{n,m}?`. Thus, `<.*?>` matches each tag separately: `<p>`, `</p>`, `<p>`, and `</p>`. In short, `.*` matches as much as possible, while `.*?` matches as little as possible, while still allowing the full pattern to succeed.
:::

#### Anchors
Anchors match a position in the text rather than a character. They are commonly used to require a match at the beginning or end of a string or line.

| Regex syntax | Matches                                |
| :----------- | :------------------------------------- |
| `^`          | The beginning of a string or line      |
| `$`          | The end of a string or line            |
| `\b`         | A word boundary                        |
| `\B`         | A position that is not a word boundary |

For example:
- `^Hello` matches `Hello` only when it appears at the beginning.
- `world$` matches `world` only when it appears at the end.
- `\bcat\b` matches the whole word `cat`, but not `catalog` or `copycat`.
- `\Bcat\B` matches `cat` only when it appears inside a larger word.

### 2.2. Advanced patterns

### Groups
A group defines several regex elements as a single unit and is written with parentheses. A group can contain a sequence such as `(https)`, alternatives such as `(jpg|png|svg)`, or an optional sequence when followed by `?`, such as `(https://)?`.

By default, a group is also a capturing group, so the regex returns both the full match and the parts matched by each group. For example, `(\d{4})-(\d{2})-(\d{2})` matches `2026-07-28` and captures `2026`, `07`, and `28` as separate components.

Capturing groups are useful when parts of a match need to be extracted, but unnecessary captures can make the output confusing. Use a non-capturing group with `(?:...)` when the group is needed only for structure.

| Purpose                       | Regex                       | Example matches                      |
| ----------------------------- | --------------------------- | ------------------------------------ |
| Optional sequence             | `(?:https://)?example\.com` | example.com, https://example.com |
| Alternative file extension    | `report.(?:pdf\|docx)`      | report.pdf, report.docx |
| Quantifier applied to a group | `(?:ha){2,4}`               | haha, hahaha, hahahaha         |


:::{note}
There are two more advanced concepts related to groups:
- *Named groups* label captured values, making results easier to read or return in a JSON-like structure.
- *Backreferences* reuse captured text later in the pattern by group index or by group name.
:::

### Lookaround
Lookarounds check whether certain text comes after or comes before a match without including that text in the match. Their negative forms check that the text is absent.

| Purpose    | Form       | Example                                  | Negative form |
| ---------- | ---------- | ---------------------------------------- | ------------- |
| Lookahead  | `(?=...)`  | `\d+(?= USD)` matches `100` in `100 USD` | `(?!...)`     |
| Lookbehind | `(?<=...)` | `(?<=\$)\d+` matches `100` in `$100`     | `(?<!...)`    |

#### Flags
| Flag            | Effect                                                 |
| --------------- | ------------------------------------------------------ |
| `re.IGNORECASE` | Makes matching case-insensitive                        |
| `re.MULTILINE`  | Makes `^` and `$` match the start and end of each line |
| `re.DOTALL`     | Makes `.` match newline characters                     |
| `re.VERBOSE`    | Allows whitespace and comments inside the pattern      |
