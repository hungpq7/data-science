# Python: Basic Concepts

## 1. Objects and variables

Everything in Python, from numbers to strings, is an object. Every object has a type, which you can inspect with the `type()` function. An assignment statement, which use the equal sign `=`, binds a variable name to an object.

```python
type(100) # int
type('python') # str
```

```python
var = 100
var = 'python'
print(var) # python
```

A variable does not require a predefined type, and the same name can later be bound to an object of another type. Python first evaluates the expression on the right-hand side of `=`, then binds the resulting object to the target on the left-hand side.

```python
a = 5
b = a
print(a, b) # bind both variables to the same object
```

```python
a = 7
a = a + 2
a
```

## 2. Operators

### 2.1. Arithmetic operators

```python
123 + 532 # addition
555 - 143 # subtraction
124 * 936 # multiplication
48 / 12   # true division
16 % 5    # remainder
16 // 5   # floor division
2 ** 5    # exponentiation
```

### 2.2. Assignment operators
Aside from basic assignment, you can also:
- Combine an arithmetic operator with `=` to perform an operation and assign the result back to the same target.
- Assign several variables in one line.

```python
x, y, z = 17, 3, 8
x += y # 20
x -= z # 12
x /= 3 # 4
x *= 5 # 20
```

### 2.3. Comparison operators
Comparison expressions evaluate to the Boolean values `True` or `False`.

```python
7 == 7
'a' != 'A'
7 > 4
2 < 5
10 >= 10
4 <= 0
```

Python allows comparisons to be chained. A chained comparison evaluates to `True` only when every comparison in the chain is true.

```python
1 < 5 < 10 != 11
(1 < 5) and (5 < 10) and (10 != 11)
```

### 3.2. Comments and line continuation
The following syntax marks are useful when writing Python code:
- A hash sign `#` starts a comment outside a string literal. Python does not execute the remainder of that physical line.
- A semicolon `;` can separate multiple simple statements on one line, although separate lines are usually easier to read.
- A backslash `\` at the end of a line explicitly continues the statement onto the next line. Python also permits cleaner implicit continuation inside `()`, `[]`, and `{}`.

```python
print("Hello, world!")
# This is a comment
```

```python
print(1); print(2)
```

```python
message = (
    'Hello, '
    'world!'
)
print(message)
```

### 3.4. Libraries
Python code can be organized into *modules* and *packages*. A module is typically a single `.py` file containing Python definitions and statements. A package groups related modules under a common namespace. The word *library* is a broader, informal term for reusable software that may contain one or more modules or packages.

Standard packages such as `os` and `datetime` are distributed with Python. Third party libraries such as `numpy` and `pandas` are developed separately from Python and need to be installed before use, usually with [`pip`]. Use `help()` to read an object's documentation and `dir()` to inspect the names available on an object, such as a module.

[PyPI]: https://pypi.org/
[`pip`]: https://pip.pypa.io/en/stable/

#### Importing a library
Use an `import` statement to make a module or selected names from a module available in the current namespace. You can:
- Import a module.
- Bind an imported module to an alias with `as`.
- Import selected names from a module.
- Bind imported names to aliases.

```python
# basic import
import math
math.degrees(math.pi)
```

```python
# import the math module under the alias m
import math as m
m.degrees(m.pi)
```

```python
# import specific functions and constants
from math import pi, degrees
degrees(pi)
```

```python
# give alias to functions and constants
from math import pi as pi_number, degrees as radian_to_degree
radian_to_degree(pi_number)
```

## 4. Control flow

### 4.1. The `if` statement
An `if` block runs when its condition is true. An optional `else` block runs when the condition is false. Indentation defines which statements belong to each block.

```python
x = 10
if x > 0:
    print('x is positive')
else:
    print('x is non-positive')
```

You can nest `if` statements, but `elif` usually provides a clearer way to express additional mutually exclusive conditions.

```python
x = 10
if x > 0:
    print('x is positive')
else:
    if x == 0:
        print('x equals 0')
    else:
        print('x is negative')
```

```python
x = 10
if x > 0:
    print('x is positive')
elif x == 0:
    print('x equals 0')
else:
    print('x is negative')
```

Python also supports single-line `if-else` expressions.

```python
x = 10
message = 'x is positive' if x > 0 else 'x is non-positive'
print(message)
```

### 4.2. The `for` loop
A `for` loop processes the items of an iterable one at a time. During each iteration, the loop variable is bound to the next item.

```python
for i in [1, 2, 3, None, 5]:
    print(i)
```

`break` exits the loop immediately; `continue` skips to the next iteration.

```python
for i in [1, 2, 3, 4, 5]:
    if i == 4:
        break
    print(i)
```

```python
total = 0
for i in [1, 2, 3, None, 5]:
    if i is None:
        continue
    total += i
total
```

### 4.3. The `while` loop
A `while` loop repeatedly executes its body while its condition remains true.

```python
x = 0
while x < 10:
    print(x, end=' ')
    x += 1
```

```python
x = 0
while True:
    x += 1
    print(x, end=' ')
    if x >= 10:
        break
```

```python
x = 0
while x < 10:
    x += 1
    if x == 5:
        continue
    print(x, end=' ')
```

### 4.4. Error handling
