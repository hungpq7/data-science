# 1.&nbsp;Estimation
Estimation là việc ước lượng các đại lượng thống kê của một tổng thể dựa trên mẫu quan sát được. Bản chất của kỹ thuật estimation là suy luận, phỏng đoán những gì chưa quan sát được nên luôn luôn đi kèm với sai số. Các phép ước lượng phổ biến trong thực tế là ước lượng khoảng giá trị của trung bình và tỷ lệ.

Để dễ nhớ, các đại lượng thống kê $\mu$ (trung bình), $p$ (tỷ lệ) mặc định là dành cho tổng thể. Các chỉ số tương ứng đo đạc trên mẫu thu thập được sẽ có thêm dấu mũ $\hat\mu,\hat{p}$ để dễ phân biệt.

## 1.1. Cấu trúc chung
Đầu ra của một estimation sẽ có dạng như sau: 90% khả năng trọng lượng hàng trung bình nằm trong khoảng từ 175 đến 185 gam. Trong đó:
- Trọng lượng hàng trung bình là **đại lượng** cần được ước lượng (ký hiệu $\theta$).
- 90% được gọi là **confidence level** - độ chính xác của ước lượng (ký hiệu $\gamma$). Giá trị này cần được chọn trước khi thực hiện estimation. Cũng có thể dùng **significance level** $\alpha=1-\gamma=10\%$ là mức sai số có thể chấp nhận.
- 175 và 185 lần lượt là chặn dưới và trên (ký hiệu $\theta_1,\theta_2$). Khoảng nằm giữa 2 giá trị này được gọi là **khoảng tin cậy** 90%.
- Khoảng tin cậy đôi khi được viết là $180\pm5$, với $e=5$ là **sai số** ước lượng.

Cấu trúc chung của một phép ước lượng được khái quát bằng công thức toán học như sau:

$$P(\theta_1<\theta<\theta_2)=1-\alpha$$

## 1.2. Ước lượng trung bình
Với một mẫu ngẫu nhiên của biến $x$, đại lượng sau được chứng minh là tuân theo t-distribution:

$$c=\frac{\hat\mu-\mu}{\hat\sigma/\sqrt{n}}\sim\text{t}(n-1)$$

<br>
<center><img src="https://raw.githubusercontent.com/hungpq7/course-resource/main/image/t_distribution.png" height=300></center>

90% giá trị của biến sẽ nằm trong khoảng từ $c_{0.05}$ đến $c_{0.95}$ (quantile 5% và 95%). Ngoài ra, vì t-distribution đối xứng nên $c_{0.05}+c_{0.95}=0$.

Nói cách khác, 90% khả năng $\hat\mu-\mu$ sẽ nằm trong khoảng:

$$c_{0.05}\frac{\hat\sigma}{\sqrt{n}}<\hat\mu-\mu<c_{0.95}\frac{\hat\sigma}{\sqrt{n}}$$

Đại lượng $e=\hat\mu-\mu$ thể hiện trung bình mẫu sai lệch bao nhiêu so với trung bình tổng thể. Do vậy, công thức estimate trung bình tổng thể được viết lại như sau:

$$
\hat{\mu}-e
<\mu<
\hat{\mu}+e
\qquad\text{for }e=-c_{\alpha/2}\frac{\hat{\sigma}}{\sqrt{n}}
$$

♈ **Coding Session 1**: Khảo sát $n=20$ hiệu sách bán cùng một cuốn sách với giá trung bình $\hat\mu=135.8$ USD và phương sai $\hat\sigma^2=23.2$. Với độ tự tin $\gamma=95\%$, giá của trung bình của cuốn sách này trên thị trường dao động trong khoảng nào?

```python
import numpy as np
import pandas as pd
from scipy import stats

alpha = 0.05
gamma = 1 - alpha
n, mean, var = 20, 135.8, 23.2

# thay vì tra bảng t value, ta dùng scipy để tính ra giá trị critical
crit = stats.t(n-1).ppf(alpha/2)
eps = -crit * np.sqrt(var/n)
lower = mean - eps
upper = mean + eps

print(f'Với khoảng tin cậy {gamma:.0%}, giá trung bình của cuốn sách nằm trong khoảng ({lower:.2f}, {upper:.2f})')
```

```python
import numpy as np
import pandas as pd
from scipy import stats

def estimate_mean(n, mu, sigma, gamma=0.95):
    alpha = 1 - gamma
    crit = stats.t(n-1).ppf(alpha/2)
    err = -crit * sigma / np.sqrt(n)
    lower = mu - err
    upper = mu + err
    print(f'Với độ tin cậy {gamma:.2%}, giá trị trung bình nằm trong khoảng ({lower:.2f}, {upper:.2f})')
    # return lower, upper
```

```python
estimate_mean(20, 135, np.sqrt(23), gamma=0.95)
```

```python
estimate_mean(n=120, mu=2.71, sigma=.43, gamma=0.9)
```

♈ **Case Study 2**: Một mẫu $n=120$ sinh viên có GPA trung bình $\hat\mu=2.71$ và độ lệch chuẩn $\hat\sigma=0.43$. Xây dựng khoảng tin cậy 90%, 95% và 99% cho GPA trung bình của toàn bộ sinh viên trong trường đó. So sánh 3 khoảng tin cậy tính được và nhận xét.

```python
gamma = 0.9
alpha = 1 - gamma
n, mean, std = 120, 2.71, 0.43

crit = stats.norm().ppf(alpha/2)
eps = -crit * std / np.sqrt(n)
lower = mean - eps
upper = mean + eps

print(f'Với {gamma:.0%} độ tin cậy, GPA trung bình nằm trong khoảng ({lower:.2f}, {upper:.2f})')
```

```python
estimate_mean(n=120, mu=2.71, sigma=.43, gamma=0.9)
```

## 1.3. Ước lượng tỷ lệ
Ước lượng này chỉ dùng cho biến binary - những biến chỉ có 2 giá trị 0 và 1. Đại lượng tỷ lệ ở đây thường là tỷ lệ số biến có giá trị 1. Rất hay gặp trong thực tế: tỷ lệ khách hàng chuyển đổi của một marketing campaign, tỷ lệ sinh viên vượt qua một bài kiểm tra,...

Công thức ước lượng tỷ lệ tổng thể $p$ khi biết tỷ lệ mẫu $\hat{p}$ như sau $c\sim\mathcal{N}(0,1)$:

$$
\hat{p}-\epsilon
<p<
\hat{p}+\epsilon
\qquad\text{for }\epsilon=-c_{\alpha/2}\sqrt{\frac{\hat{p}(1-\hat{p})}{n}}
$$

♈ **Coding Session 3**: Để kiểm tra chất lượng hàng hoá trong một container, người ta kiểm tra ngẫu nhiên $n=120$ gói hàng và phát hiện ra 5 gói bị hỏng. Với độ tin cậy 95%, tỷ lệ hàng hoá đạt chất lượng nằm trong khoảng nào?

```python
gamma = 0.95
alpha = 1 - gamma
n = 120
p = 1 - 5/n

crit = stats.norm().ppf(alpha/2)
err = -crit * np.sqrt(p * (1-p) / n)
lower = p - err
upper = p + err

print(f'Với độ tin cậy {gamma:.0%}, tỷ lệ hàng lỗi của container nằm trong khoảng ({lower:.2%}, {upper:.2%})')
```

Cũng tương tự ước lượng trung bình tổng thể, công thức ước lượng tỷ lệ tổng thể cũng có thể mở rộng để so sánh 2 tỷ lệ với nhau.

$$
(\hat{p}_1-\hat{p}_2)-\epsilon
<p_1-p_2<
(\hat{p}_1-\hat{p}_2)+\epsilon
\qquad\text{for }\epsilon=-c_{\alpha/2}\sqrt{\frac{\hat{p}_1(1-\hat{p}_1)}{n_1}+\frac{\hat{p}_2(1-\hat{p}_2)}{n_2}}
$$

♈ **Coding Session 4**: Steve muốn mua cuốn sách "The Art of Probability" trên Amazon và tham khảo được 2 nơi bán như sau:
- Shop GreenEggs có $n_1=50$ ratings với tỷ lệ đánh giá tích cực là $\hat p_1=96\%$
- Shop HammingTime có $n_2=200$ ratings với tỷ lệ đánh giá tích cực là $\hat p_2=93\%$

Với mức sai số cho phép $\alpha=10\%$, hãy giúp Steve đưa ra quyết định nên chọn mua sách ở shop nào.

```python
import numpy as np
from scipy import stats

n1, n2 = 50, 200
p1 = 0.96
p2 = 0.93
alpha = 0.1
gamma = 1 - alpha

crit = stats.norm().ppf(alpha/2)
eps = -crit * np.sqrt(p1 *(1-p1)/n1 + p2*(1-p2)/n2)
lower = p1 - p2 - eps
upper = p1 - p2 + eps

print(f'Với độ tin cậy {gamma:.0%}, chênh lệch tỷ lệ review tích cực nằm trong khoảng ({lower:.2%}, {upper:.2%})')
```

# 2.&nbsp;Linear regression
Hồi quy tuyến tính (đơn biến, đa biến) là phương pháp cơ bản nhưng rất hữu dụng. Là nền tảng để phát triển các mô hinh tuyến tính trong học máy và kinh tế lượng. Người ta đã nghiên cứu ra rằng rất nhiều mối quan hệ của các sự kiện trong đời sống và kinh tế là mối quan hệ tuyến tính. Đo lường được mối quan hệ này có thể giúp chúng ta điều chỉnh các yếu tố sao cho kết quả ra tối ưu nhất.

Trong thống kê, hồi quy tuyến tính tuân theo nhiều giả thuyết. Khác với Machine Learning, thống kê đề cao ý nghĩa của các hệ số - do các hệ số đó phản ánh mối quan hệ giữa các biến độc lập với biến phụ thuộc

## 2.1. Mô hình
Linear Regression là kỹ thuật dùng để mô phỏng mối quan hệ giữa các biến giải thích ($x_1,x_2,\dots$) với biến mục tiêu ($y$). VD, giá nhà (biến mục tiêu) bị ảnh hưởng bởi các yếu tố như diện tích, số phòng, khoảng cách tới trung tâm,... (các biến giải thích). Mối quan hệ này được mô tả bởi phương trình toán học sau:

$$\hat{y}=\beta_0+\beta_1x_1+\beta_2x_2+\beta_3x_3+\ldots$$

Phương trình này đơn giản là 1 đường thẳng (nếu chỉ có duy nhất 1 biến giải thích) hoặc 1 mặt phẳng (nếu có đúng 2 biến giải thích), vì vậy mô hình này được gọi là tuyến tính. Để dễ hình dung, ta sẽ tìm hiểu cách xây dựng mô hình đơn biến: $\hat{y}=a+bx$.
<br>

<center><img src="https://raw.githubusercontent.com/hungpq7/course-resource/main/image/linear_regression.svg" height=300></center>

Trong phương trình này $a$ là hệ số chặn và không mang nhiều ý nghĩa. $b$ là hệ số góc và thể hiện mức độ ảnh hưởng của $x$ lên $y$, cụ thể:
- $b<0$ thì ảnh hưởng nghịch biến, $b>0$ thì ảnh hưởng đồng biến
- $|b|$ càng lớn thì ảnh hưởng của $x$ lên $y$ càng lớn.

**Animation**: https://www.geogebra.org/calculator/hr4mnxdf

Các hệ số $a,b$ được tìm ra thế nào? Phương pháp **OLS** (Ordinary Least Squares) giải quyết bài toán này bằng việc tìm $a,b$ sao cho tổng bình phương sai số (**SSE** - Sum of Squared Errors) là nhỏ nhất:

$$\text{SSE}=\sum_{n=1}^{N}(y_n-\hat{y}_n)^2$$

Việc tính toán ra các hệ số $a,b$ thoả mãn tính chất trên đã được các phần mềm như Python, R, SPSS tích hợp sẵn và chúng ta không cần phải giải lại. Quá trình giải ra các hệ số này được gọi là chạy mô hình (Statistics) hay huấn luyện mô hình (Machine Learning).

## 2.2. Hướng tiếp cận Statistics

### Mục đích
Linear Regression trong Statistics được coi là cách tiếp cận truyền thống, tập trung vào việc biểu diễn **mối quan hệ** giữa $x$ và $y$. Cụ thể, mô hình sẽ cho đầu ra là các insight như:
- Mỗi căn nhà cứ xa trung tâm thêm 1km thì giá sẽ giảm 200 triệu.
- Diện tích nhà cứ tăng thêm 1m2 thì giá sẽ tăng 50 triệu.

### Giả định
Để các insight như vậy đáng tin cậy, mô hình cần thoả mãn rất nhiều (từ 5 đến 8, tuỳ tài liệu) giả định, trong đó tiêu biểu nhất là:
- Không có hiện tượng đa cộng tuyến (multi-collinearity), là hiện tượng mà 2 biến giải thích tương quan rất mạnh với nhau
- Sai số ($\epsilon=y-\hat{y}$) tuân theo phân phối chuẩn với trung bình là 0
- Biến mục tiêu ($y$) không tương quan với sai số ($\epsilon$).
- Phương sai của $\epsilon$ phải nhất quán, không thay đổi nhiều khi xét các mức khác nhau của biến giải thích.

Để kiểm tra các giả định này có thoả mãn hay không, có 2 cách là (1) vẽ các biểu đồ cần thiết và kiểm tra cảm tính, (2) kiểm định giả thuyết thống kê. Chỉ cần có 1 giả định không thoả mãn, mô hình sẽ được coi là không đáng tin cậy.

### Các bước
Triển khai mô hình Linear Regression là một vòng lặp các bước thử sai như sau:
- Chạy mô hình
- Kiểm tra tất cả giả định xem đã thoả mãn hay chưa
- Nếu chưa, biến đổi hoặc thêm bớt biến giải thích (dựa vào kinh nghiệm)
- Tiếp tục chạy lại mô hình

## 2.3. Hướng tiếp cận Machine Learning
Machine Learning coi Linear Regression như một kỹ thuật để **dự đoán** biến mục tiêu, với tiêu chí là dự đoán $y$ tốt nhất trên những dữ liệu mà mô hình chưa nhìn thấy bao giờ. Do vậy thay vì tìm cách thoả mãn các giả định, Machine Learning sử dụng cách tiếp cận rất đơn giản và thực tế:

> Tối ưu **evaluation metric** (chỉ số đánh giá) trên dữ liệu mới



### Chỉ số đánh giá
VD: 3 ngôi nhà trên thực tế có giá 4.9 tỷ, 5.3 tỷ và 7.2 tỷ, 2 mô hình dự đoán như sau:
- Mô hình 1 đoán cả 3 căn nhà có giá 5.8 tỷ.
- Mô hình 2 dự đoán lần lượt 4.7 và 5.2 tỷ 7.5 tỷ.

Về mặt cảm quan, dễ dàng thấy mô hình 2 tốt hơn hẳn. Nhưng ta cần 1 thước đo cụ thể, lượng hoá được. Có 2 evaluation metric phổ biến như sau.

**RMSE** (Root Mean Squared Error), chạy từ $0$ đến $+\infty$. Gắn liền với đơn vị của biến dự đoán, VD: giá dự đoán 100 căn nhà có RMSE là 92 triệu.

$$\mbox{MSE} = \mbox{RMSE}^2 = {\frac{1}{N}} \sum_{n=1}^{N}(y_n-\hat{y}_n)^{2}$$

**R2** (R-squared/Coefficient of Determination), chạy từ $-\infty$ đến $1$. Càng gần 1 càng tốt, nhưng $>0.3$ là chấp nhận được.

$$R^2=1-\dfrac{\text{MSE}}{\sigma_y^2}=1-\dfrac{\text{MSE}_\text{model}}{\text{MSE}_\text{baseline}}$$

Làm thế nào để biết rằng các metric như RMSE hay R2 là tốt nhất trên dữ liệu hoàn toàn mới?

```python
import numpy as np
np.set_printoptions(precision=4, suppress=True)
import pandas as pd
from scipy import stats

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score as R2, mean_squared_error as MSE
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/hungpq7/course-resource/main/data/boston.csv')
df.head()
```

```python
x = df.drop(columns='price')
y = df['price']

xTrain, xTest, yTrain, yTest = train_test_split(x, y, test_size=0.2, random_state=123)
```

```python
model = LinearRegression()
model.fit(xTrain, yTrain)
```

```python
model.coef_
```

```python
yPred = model.predict(xTest)
```

```python
R2(yTest, yPred)
```

```python
MSE(yTest, yPred)
```
